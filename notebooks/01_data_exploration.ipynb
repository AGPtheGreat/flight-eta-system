```python
from pathlib import Path
import zipfile

data_path = Path("../data/raw/flights.csv")
raw_dir = Path("../data/raw")

# Check that the downloaded file is a ZIP archive
print("Is ZIP file:", zipfile.is_zipfile(data_path))

# List the files inside the archive
with zipfile.ZipFile(data_path, "r") as archive:
    print("Files inside archive:")
    for name in archive.namelist():
        print(name)

    # Extract the CSV into data/raw/
    archive.extractall(raw_dir)

print("Extraction complete!")
```

```python
from pathlib import Path
import sys
import pandas as pd

# Make the project's src directory importable.
project_root = Path.cwd().parent
sys.path.insert(0, str(project_root / "src"))

from data.processor import clean_flight_data, save_clean_flight_data

# Load the extracted raw data.
raw_path = project_root / "data" / "raw" / "T_ONTIME_REPORTING.csv"

flights = pd.read_csv(raw_path)

# Clean the data.
cleaned_flights = clean_flight_data(flights)

print("Original records:", len(flights))
print("Cleaned records:", len(cleaned_flights))
print("Records removed:", len(flights) - len(cleaned_flights))

# Save processed data.
output_path = project_root / "data" / "processed" / "flights_clean.csv"
save_clean_flight_data(cleaned_flights, str(output_path))

print("Saved cleaned data to:", output_path)
cleaned_flights.head()
```
