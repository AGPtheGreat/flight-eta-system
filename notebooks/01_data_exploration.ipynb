```python
# Calculates mean absolute error (MAE) and root mean squared error (RMSE) for the 
# baseline model that predicts scheduled duration.

from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Sort by date to keep the evaluation chronological.
data = cleaned_flights.sort_values("FL_DATE").copy()

# Split by distinct dates: earlier dates for training,
# later dates for testing.
dates = sorted(data["FL_DATE"].unique())
split_index = int(len(dates) * 0.8)
cutoff_date = dates[split_index]

train = data[data["FL_DATE"] < cutoff_date]
test = data[data["FL_DATE"] >= cutoff_date]

# Baseline: predict scheduled duration.
y_true = test["ACTUAL_ELAPSED_TIME"]
y_pred = test["CRS_ELAPSED_TIME"]

mae = mean_absolute_error(y_true, y_pred)
rmse = np.sqrt(mean_squared_error(y_true, y_pred))

print("Training flights:", len(train))
print("Testing flights:", len(test))
print("Test period starts:", cutoff_date)
print(f"Baseline MAE: {mae:.2f} minutes")
print(f"Baseline RMSE: {rmse:.2f} minutes")
```

```python
# Building the model:
#
# Using a Random Forest Regressor for these reasons:
# 
# It can learn nonlinear relationships between features.
# It can identify interactions between airlines, airports, and departure times.
```

```python
# Create copies so we don't modify the original datasets.
train_ml = train.copy()
test_ml = test.copy()

# Extract departure hour from HHMM format.
train_ml["DEPARTURE_HOUR"] = train_ml["CRS_DEP_TIME"] // 100
test_ml["DEPARTURE_HOUR"] = test_ml["CRS_DEP_TIME"] // 100

# Correct any invalid hour values defensively.
train_ml["DEPARTURE_HOUR"] = train_ml["DEPARTURE_HOUR"].where(
    train_ml["DEPARTURE_HOUR"] < 24, 0
)

test_ml["DEPARTURE_HOUR"] = test_ml["DEPARTURE_HOUR"].where(
    test_ml["DEPARTURE_HOUR"] < 24, 0
)

# Extract day of week (Monday = 0, Sunday = 6).
train_ml["DAY_OF_WEEK"] = train_ml["FL_DATE"].dt.dayofweek
test_ml["DAY_OF_WEEK"] = test_ml["FL_DATE"].dt.dayofweek

# Define model features.
feature_columns = [
    "OP_UNIQUE_CARRIER",
    "ORIGIN",
    "DEST",
    "DEPARTURE_HOUR",
    "DAY_OF_WEEK",
    "CRS_ELAPSED_TIME"
]

X_train = train_ml[feature_columns]
X_test = test_ml[feature_columns]

y_train = train_ml["ACTUAL_ELAPSED_TIME"]
y_test = test_ml["ACTUAL_ELAPSED_TIME"]

print("Training feature dimensions:", X_train.shape)
print("Testing feature dimensions:", X_test.shape)

X_train.head()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor

# Identify categorical and numerical features.
categorical_features = [
    "OP_UNIQUE_CARRIER",
    "ORIGIN",
    "DEST"
]

numerical_features = [
    "DEPARTURE_HOUR",
    "DAY_OF_WEEK",
    "CRS_ELAPSED_TIME"
]

# Convert categorical variables into numerical columns.
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)

# Create the Random Forest model.
random_forest = RandomForestRegressor(
    n_estimators=80,
    max_depth=20,
    min_samples_leaf=5,
    n_jobs=-1,
    random_state=42
)

# Combine preprocessing and model training.
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", random_forest)
    ]
)

print("Model pipeline created successfully!")
```

```python
print("Training Random Forest model...")

model.fit(X_train, y_train)

print("Training complete!")
```

```python
# Code that evaluates the Random Forest model and compares it to the baseline shown above.

from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Generate predictions.
predictions = model.predict(X_test)

# Calculate evaluation metrics.
rf_mae = mean_absolute_error(y_test, predictions)

rf_rmse = np.sqrt(
    mean_squared_error(y_test, predictions)
)

print(f"Random Forest MAE: {rf_mae:.2f} minutes")
print(f"Random Forest RMSE: {rf_rmse:.2f} minutes")

# Compare against baseline.
print("\nModel comparison:")
print(f"Baseline MAE:     {mae:.2f} minutes")
print(f"Random Forest MAE: {rf_mae:.2f} minutes")

improvement = ((mae - rf_mae) / mae) * 100

print(f"MAE improvement: {improvement:.2f}%")
```

```python
# Calculate the absolute difference between predicted and actual duration.
absolute_errors = np.abs(y_test - predictions)

# Calculate the percentage of predictions falling within each error threshold.
thresholds = [5, 10, 15, 20, 30]

print("Prediction accuracy by error threshold:\n")

for threshold in thresholds:
    percentage = (absolute_errors <= threshold).mean() * 100

    print(
        f"Within {threshold:2d} minutes: "
        f"{percentage:.2f}% of predictions"
    )
```

```python
# Create a separate DataFrame to compare predictions with actual results.
error_analysis = test_ml[
    [
        "FL_DATE",
        "OP_UNIQUE_CARRIER",
        "ORIGIN",
        "DEST",
        "CRS_ELAPSED_TIME",
        "ACTUAL_ELAPSED_TIME"
    ]
].copy()

# Attach the model's predictions and calculate the signed and absolute errors.
error_analysis["PREDICTED_DURATION"] = predictions

error_analysis["ERROR"] = (
    error_analysis["PREDICTED_DURATION"]
    - error_analysis["ACTUAL_ELAPSED_TIME"]
)

error_analysis["ABSOLUTE_ERROR"] = error_analysis["ERROR"].abs()

# Display the 20 flights with the largest prediction errors.
largest_errors = error_analysis.sort_values(
    "ABSOLUTE_ERROR",
    ascending=False
).head(20)

largest_errors
```

```python
# Group flights according to their actual duration.
duration_bins = [0, 60, 120, 180, 240, 360, float("inf")]

duration_labels = [
    "Under 1 hour",
    "1-2 hours",
    "2-3 hours",
    "3-4 hours",
    "4-6 hours",
    "Over 6 hours"
]

error_analysis["DURATION_GROUP"] = pd.cut(
    error_analysis["ACTUAL_ELAPSED_TIME"],
    bins=duration_bins,
    labels=duration_labels,
    right=False
)

# Calculate average prediction error for each duration group.
duration_performance = error_analysis.groupby(
    "DURATION_GROUP",
    observed=False
).agg(
    FLIGHT_COUNT=("ABSOLUTE_ERROR", "count"),
    MAE=("ABSOLUTE_ERROR", "mean")
)

duration_performance
```
